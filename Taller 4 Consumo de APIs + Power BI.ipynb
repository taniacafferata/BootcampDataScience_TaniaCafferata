## TALLER APIs


```python
import pandas as pd
import requests

# 1. Consulta de la API

url = "https://www.datos.gov.co/api/v3/views/cuit-be64/query.json"
respuesta = requests.get(url)

if respuesta.status_code == 200:
    print("Código de estado:", respuesta.status_code)

    datos_json = respuesta.json()

    df_datos = pd.DataFrame(datos_json)

    display(df_datos.head())
else:
    print(
        f"Error al conectar con Datos Abiertos. Código de estado: {respuesta.status_code}"
    )
```

```python
## Limpieza de datos
  ## Eliminar columnas

columnas_a_eliminar = [':id', ':version', ':created_at', ':updated_at']
df_limpio = df_datos.drop(columns=columnas_a_eliminar, errors='ignore')

# Renombrar columnas 
df_limpio = df_limpio.rename(columns={
    'a_o': 'Año',
    'razon_social': 'Empresa',
    'no_suscriptores_acueducto': 'Suscriptores Acueducto',
    'consumo_m3_acueducto':'Consumo M3 Agua',
    'promedio_consumo_acueducto': 'Promedio Consumo Agua',
    'no_suscriptores_alcantarillado':'Suscriptores Alcantarillado',
    'consumo_m3_alcantarillado':'Consumo M3 Alcantarillado',
    'promedio_consumo_alcantarillado':'Promedio Consumo Alcantarillado'
})

## Eliminar la palabra estrato en los datos
if 'estrato' in df_limpio.columns:
    df_limpio['estrato'] = df_limpio['estrato'].str.replace('Estrato', '', case=False).str.strip()
```

```python
display(df_limpio.head())
```

```python
## Exportar database en un archivo .CSV

nombre_archivo = "dataset_acueducto_alcantarillado.csv"

df_limpio.to_csv(nombre_archivo, index=False, encoding="utf-8-sig")

print(f"\n¡Dataset procesado con éxito!")
print(f"Archivo guardado en tu directorio local como: '{nombre_archivo}'")


display(df_limpio.head())
```
